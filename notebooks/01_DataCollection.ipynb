```python
!pip install yfinance
!pip install finmind
!pip install pandas

import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt

from pathlib import Path
from datetime import datetime
```

```python
PROJECT_ROOT = Path("/content/drive/MyDrive/TAIF")

RAW_DATA = PROJECT_ROOT / "data" / "raw"

CHART_PATH = PROJECT_ROOT / "data" / "charts"

RAW_DATA.mkdir(parents=True, exist_ok=True)
CHART_PATH.mkdir(parents=True, exist_ok=True)

print(PROJECT_ROOT)
```

```python
ticker = "2330.TW"

start_date = "2024-01-01"

end_date = "2025-01-01"

stock = yf.Ticker(ticker)

df = stock.history(
    start=start_date,
    end=end_date
)

df.head()
```

```python
print("="*50)

print("Data Information")

print("="*50)

print(df.info())

print()

print(df.isnull().sum())
```

```python
df.head()
df.tail()
```

```python
plt.figure(figsize=(12,5))

plt.plot(df.index, df["Close"])

plt.title("2330.TW Close Price")

plt.xlabel("Date")

plt.ylabel("Price")

plt.grid(True)

plt.show()
```

```python
file_name = "raw_2330_2024.csv"

df.to_csv(
    RAW_DATA / file_name,
    encoding="utf-8-sig"
)

print(f"Saved : {file_name}")
```

```python
plt.figure(figsize=(12,5))

plt.plot(df.index, df["Close"])

plt.title("2330.TW Close Price")

plt.grid(True)

plt.savefig(
    CHART_PATH / "2330_close.png",
    dpi=300,
    bbox_inches="tight"
)

plt.close()

print("Chart Saved")
```
